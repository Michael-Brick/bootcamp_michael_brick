## Preprocessing Assumptions and Reflection

`fill_missing_median()` and `drop_missing()` were applied as part of
the preprocessing workflow but did not alter the dataset, because the raw dataset contained no missing values. No values needed to be imputed and no observations were removed.

The `date` column was converted from text to a datetime data type.

The `value` column was normalized using MinMaxScaler, placing the
observations on a common scale from 0 to 1 while preserving their
ordering and proportional distances.  This was appropriate considering
that the dataset contained no extreme outliers.

The cleaned dataset retains all 10 original observations. The primary
changes are the corrected date type and normalization of the numeric
`value` feature.

Median imputation would preserve observations if numeric values were
missing, but it could reduce natural variation in the data. Dropping
missing observations avoids creating artificial values but can reduce
sample size or introduce bias if the missingness is systematic.
Neither tradeoff affected this dataset because no values were missing.
